```python
import pandas as pd
import numpy as np
import os

# Funções básicas para carregar e inspecionar a base de cadastro

def carregar_e_mostrar(caminho_arquivo, n=5):
    """Carrega um arquivo Excel para um DataFrame do pandas e mostra as primeiras linhas.

    Args:
        caminho_arquivo (str): caminho para o arquivo Excel.
        n (int): número de linhas a exibir com `head()` (padrão 5).

    Returns:
        pandas.DataFrame: o DataFrame carregado.
    """
    # Lê o arquivo Excel para um DataFrame
    df = pd.read_excel(caminho_arquivo)
    # Mostra as primeiras n linhas para inspecionar os dados rapidamente
    display(df.head(n))
    return df

# Exemplo: carrega o arquivo local "base_malu.xlsx" e mostra 5 linhas
df_cadastro = carregar_e_mostrar("base_malu.xlsx", 5)

def base_descricao(df):
    """Gera estatísticas descritivas básicas para colunas numéricas.

    Usa `df.describe()` do pandas, que inclui contagem, média, desvio padrão,
    mínimos, máximos e quartis. Serve para identificar rapidamente a distribuição
    e potenciais valores inconsistentes.
    """
    descricao = df.describe()
    display(descricao)
    return descricao

descricao_cadastro = base_descricao(df_cadastro)

def base_info(df):
    """Mostra informações gerais do DataFrame.

    `df.info()` apresenta o tipo de cada coluna, quantos valores não nulos existem
    e estimativa de memória usada — útil para detectar colunas com muitos nulos
    ou tipos inesperados (por exemplo, números carregados como strings).
    """
    info = df.info()
    display(info)
    return info

info_cadastro = base_info(df_cadastro)

def exibir_colunas(df):
    """Retorna os nomes das colunas do DataFrame.

    Útil para decidir transformações a aplicar (renomear, excluir, converter tipos).
    """
    colunas = df.columns
    display(colunas)
    return colunas

colunas_cadastro = exibir_colunas(df_cadastro)
```

```python
# Função para remover uma coluna do DataFrame
def excluir_coluna(df, coluna):
    """Remove uma coluna e retorna um novo DataFrame sem alterara o original.

    Args:
        df (pandas.DataFrame): DataFrame de entrada.
        coluna (str): nome da coluna a remover.

    Returns:
        pandas.DataFrame: DataFrame resultante sem a coluna especificada.
    """
    # `drop` com `columns=[coluna]` cria um novo DataFrame sem a coluna
    df_base = df.drop(columns=[coluna])
    # Exibir as primeiras linhas ajuda a verificar se a exclusão ocorreu corretamente
    display(df_base.head())
    return df_base

# Exemplo: nome da coluna que será removida (ajuste conforme necessário)
coluna_para_excluir = "OBS"

df_base = excluir_coluna(df_cadastro, coluna_para_excluir)
```

```python
# Função para padronizar/renomear colunas com nomes mais legíveis
def corrigir_colunas(df):
    """Renomeia colunas usando um mapeamento explícito.

    Observações:
    - É comum ter nomes originais em caixa alta ou abreviados; renomear deixa
      o dataset mais compreensível para análise e relatórios.
    - Se uma chave não existir no DataFrame, `rename` simplesmente a ignora.
    """
    mapeamento = {
        "RAZAO_SOCIAL": "Razão Social",
        "CD_AREA_RGST": "Area de Registro",
        "TERMINAL": "Terminal",
        "NR_TLFN": "Telefone",
        "NOME_GESTOR": "Gestor",
        "BAIRRO": "Bairro",
        "TLF1_GESTOR": "Telefone do Gestor",
        "EMAIL_GESTOR": "Email do Gestor",
        "CEP": "CEP",
        "LOGRADOURO": "Logradouro",
        "QT_LINHAS_TOTAL": "Total Linhas",
        "NR_ENDR_NMRO": "Número do Endereço",
        "DS_ENDR_CMPL_1": "Complemento",
        "NR_CEP": "CEP",
        "DS_CDDE": "Município",
        "NR_ENDR_NMRO": "N°",
        'CONSULTORA': 'Consultora',
        'AGENDOU (Sim ou Não)': 'Agendado',
        'DATA': 'Data',        
        "EMAIL": "Email"
    }

    # Aplicar o mapeamento de nomes ao DataFrame
    df_corrigido = df.rename(columns=mapeamento)
    # Mostrar para conferir
    display(df_corrigido.head())
    return df_corrigido


df_corrigido = corrigir_colunas(df_base)
```

```python
# Função para aplicar formatações por linha/coluna (exemplos comuns)
def formatar_linhas(df):
    """Aplica transformações que tornam os dados consistentes para análise.

    Exemplos típicos:
    - padronizar datas
    - normalizar formatos de telefone
    - tratar strings vazias ou valores nulos
    """
    # Converter a coluna 'Data' para formato 'dd/mm/yyyy' quando presente
    if 'Data' in df.columns:
        # `errors='coerce'` transforma valores inválidos em NaT (facilita limpeza)
        df['Data'] = pd.to_datetime(df['Data'], errors='coerce').dt.strftime('%d/%m/%Y')

    display(df.head())
    return df

df_formatado = formatar_linhas(df_corrigido)

# Reaproveita as funções de inspeção para o DataFrame formatado
descricao_formatada = base_descricao(df_formatado)
info_formatada = base_info(df_formatado)
```

```python
def data_tipo(df):
    """Converte a coluna 'Data' para o tipo datetime do pandas.

    Usar tipos datetime permite ordenar por data, filtrar por intervalo e extrair
    componentes (ano, mês, dia) com facilidade.
    """
    # Informar `format` acelera a conversão quando o formato é conhecido.
    df["Data"] = pd.to_datetime(df["Data"], format='%d/%m/%Y', errors='coerce')    
    display(df.dtypes)
    return df

# Converte e atualiza o DataFrame formatado
df_formatado = data_tipo(df_formatado)
```

```python
#padronização de dados
def padronizar_dados(df):
    """Aplica uma série de transformações para padronizar os dados.

    Combina as etapas anteriores (exclusão de colunas, renomeação, formatação)
    em um processo único e reutilizável. Ideal para garantir que o dataset esteja
    consistente antes de análises ou visualizações.
    """
    df["Razão Social"] = df["Razão Social"].str.title()  # converte os valores da coluna "Razão Social" para título (primeira letra maiúscula)
    
    df["Gestor"] = df["Gestor"].str.title()  # converte os valores da coluna "Gestor" para título (primeira letra maiúscula)
    
    df["Bairro"] = df["Bairro"].str.title()  # garante que o nome do bairro esteja em título
    
    df["Logradouro"] = df["Logradouro"].str.title()  # padroniza o logradouro para formato de título
    
    df["Município"] = df["Município"].str.title()  # padroniza o nome do município em título
    
    df["Consultora"] = df["Consultora"].str.title()  # padroniza o nome da consultora em título
    
    df["Email do Gestor"] = df["Email do Gestor"].str.lower()  # converte os valores da coluna "Email do Gestor" para minúsculas
    
    # Exibir as primeiras linhas para verificar a padronização
    display(df.head(2))
    
    return df
df_padronizado = padronizar_dados(df_formatado)
```

```python
#tratamento de valores nulos
def tratar_valores_nulos(df):
    """Preenche ou remove valores nulos para garantir consistência.

    Estratégias comuns:
    - preencher com um valor padrão (ex: "Desconecido", 0, média)
    - remover linhas ou colunas com muitos nulos
    - usar técnicas de imputação mais avançadas (ex: KNN, regressão)
    """
    # Exemplo simples: preencher valores nulos nas colunas a seguir
    df['Bairro'] = df['Bairro'].fillna('Desconecido')
    df['Consultora'] = df['Consultora'].fillna('Sem Consultora')
    df['Agendado'] = df['Agendado'].fillna('Sem Agendamento')
    df['Complemento'] = df['Complemento'].fillna('Sem Complemento')
    df['Data'] = df['Data'].fillna('Sem Data')
    
    display(df.head(100))
    return df
#chamada da função para tratar valores nulos
df_tratado = tratar_valores_nulos(df_padronizado)
```

```python
#tratamento de valores duplicados
def tratar_valores_duplicados(df):
    """Remove ou marca linhas duplicadas para evitar distorções na análise.

    Estratégias comuns:
    - `drop_duplicates()` para remover linhas idênticas
    - `duplicated()` para marcar quais linhas são duplicatas
    - Analisar quais colunas definem a duplicidade (ex: mesmo CNPJ, mesmo endereço)
    """
    # Exemplo: remover linhas completamente idênticas
    df_sem_duplicados = df.drop_duplicates()
    
    display(df_sem_duplicados.head(3))
    return df_sem_duplicados

# Aplica a limpeza definitiva de duplicados ao DataFrame tratado
df_final = tratar_valores_duplicados(df_tratado)
```

```python
#exportar o DataFrame final para um novo arquivo Excel
def exportar_para_excel(df, caminho_arquivo):
    """Salva o DataFrame em um arquivo Excel."""
    df.to_excel(caminho_arquivo, index=False)
    print(f"DataFrame exportado com sucesso para {caminho_arquivo}")

# Exemplo de uso
# Caminho de destino para salvar o resultado final em Excel
caminho = r"C:\Users\XXX\Desktop\data\df_final.xlsx"
exportar_para_excel(df_final, caminho)
```
